```python
import pandas as pd
import os
```

```python
data = {
    "Employee_ID": [101,102,103,104,105,106,107,108,109,110],
    "Name": ["Arjun","Priya","Rahul","Sneha","Amit","Kavya","Rohan","Neha","Vikas","Anjali"],
    "Department": ["IT","HR","Finance","IT","Sales","HR","Finance","Sales","IT","Finance"],
    "City": ["Hyderabad","Bengaluru","Chennai","Delhi","Mumbai","Hyderabad","Pune","Delhi","Mumbai","Chennai"],
    "Experience": [2,5,4,3,6,7,5,4,2,6],
    "Salary": [45000,65000,60000,50000,70000,75000,62000,68000,48000,64000]
}

df = pd.DataFrame(data)

df
```

```python
print("Shape:", df.shape)

print("\nData Types:")
print(df.dtypes)

print("\nFirst 10 Rows:")
df.head(10)
```

```python
# Filter employees with salary above 60000

filtered_df = df[df["Salary"] > 60000]

print("Filtered Data:")
display(filtered_df)


# Group employees by department and calculate average salary

group_df = df.groupby("Department")["Salary"].mean().reset_index()

print("Department-wise Average Salary:")
display(group_df)
```

```python
# Creating department manager dataset

manager_df = pd.DataFrame({
    "Department": ["IT","HR","Finance","Sales"],
    "Manager": ["Rajesh","Suman","Kiran","Meena"]
})


# Merge employee data with manager data

merged_df = pd.merge(df, manager_df, on="Department")

merged_df
```

```python
# Department and city wise average salary

pivot_df = pd.pivot_table(
    df,
    values="Salary",
    index="City",
    columns="Department",
    aggfunc="mean"
)

pivot_df
```

```python
# Export files

df.to_csv("cleaned_employee.csv", index=False)

df.to_parquet("cleaned_employee.parquet", index=False)


# Compare file sizes

csv_size = os.path.getsize("cleaned_employee.csv")

parquet_size = os.path.getsize("cleaned_employee.parquet")


print("CSV File Size:", csv_size, "bytes")

print("Parquet File Size:", parquet_size, "bytes")
```

```python
## Observation

- Loaded and analyzed the Indian employee dataset using Pandas DataFrame.
- Checked dataset shape, column data types, and displayed the first 10 records.
- Applied filtering to extract employees based on salary conditions.
- Used GroupBy to calculate department-wise average salary.
- Performed Merge operation to combine employee data with department manager details.
- Created a Pivot Table to summarize salary distribution across departments and cities.
- Exported the cleaned dataset into CSV and Parquet formats.
- Compared file sizes of CSV and Parquet files and observed the difference in storage formats.

Conclusion:
Pandas provides efficient tools for data manipulation, aggregation, joining, and exporting datasets. These operations are useful for preparing data before analysis and machine learning workflows.
```
