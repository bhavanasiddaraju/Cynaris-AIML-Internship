```python
import requests
import json

OLLAMA_URL = "http://127.0.0.1:11434/api/chat"

print("Libraries imported successfully")
```

```python
def ask_ollama(prompt, system_prompt="You are a helpful AI assistant."):
    payload = {
        "model": "llama3.2:3b",
        "messages": [
            {
                "role": "system",
                "content": system_prompt
            },
            {
                "role": "user",
                "content": prompt
            }
        ],
        "stream": False
    }

    response = requests.post(OLLAMA_URL, json=payload)
    response.raise_for_status()

    result = response.json()
    return result["message"]["content"]

print("Ollama API function created successfully")
```

```python
response = ask_ollama(
    "What is artificial intelligence?",
    "You are an AI teacher. Explain concepts clearly for a college student."
)

print(response)
```

```python
prompts = [
    "What is machine learning?",
    "Explain neural networks in simple terms.",
    "What is the difference between AI and machine learning?",
    "Give three real-world applications of large language models.",
    "Explain overfitting in machine learning."
]

system_prompt = """
You are a helpful AI/ML tutor.
Give clear, concise and easy-to-understand answers
for an engineering college student.
"""

for i, prompt in enumerate(prompts, 1):
    print(f"\n--- Prompt {i} ---")
    print(prompt)
    print("\nResponse:")
    print(ask_ollama(prompt, system_prompt))
```

```python
def ask_model(model, prompt, system_prompt="You are a helpful AI assistant."):
    payload = {
        "model": model,
        "messages": [
            {
                "role": "system",
                "content": system_prompt
            },
            {
                "role": "user",
                "content": prompt
            }
        ],
        "stream": False
    }

    response = requests.post(OLLAMA_URL, json=payload)
    response.raise_for_status()

    return response.json()["message"]["content"]
```
