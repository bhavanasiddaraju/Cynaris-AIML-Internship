```python
import ollama

def ask_ollama(prompt):
    response = ollama.chat(
        model="llama3.2:3b",
        messages=[
            {
                "role": "system",
                "content": "You are a helpful AI assistant. Give clear, accurate and concise answers."
            },
            {
                "role": "user",
                "content": prompt
            }
        ]
    )

    return response["message"]["content"]
```

```python
prompts = [
    "What is machine learning?",
    "Explain neural networks in simple terms.",
    "What is RAG in generative AI?",
    "Why is data preprocessing important?",
    "What are the advantages of local LLMs?"
]

for i, prompt in enumerate(prompts, 1):
    print(f"\n--- Prompt {i} ---")
    print(prompt)
    print("\nResponse:")
    print(ask_ollama(prompt))
```

```python
questions = [
    "Explain artificial intelligence to a beginner.",
    "What are the main benefits of using generative AI?",
    "Explain the difference between supervised and unsupervised learning."
]

models = ["llama3.2:3b", "qwen2.5:3b"]

for question in questions:
    print("\n" + "="*70)
    print("QUESTION:", question)

    for model in models:
        response = ollama.chat(
            model=model,
            messages=[
                {
                    "role": "system",
                    "content": "You are a helpful AI assistant. Give clear, accurate and easy-to-understand answers."
                },
                {
                    "role": "user",
                    "content": question
                }
            ]
        )

        print(f"\n--- {model} ---")
        print(response["message"]["content"])
```

```python
comparison = """
Model Comparison:

llama3.2:3b:
- Produced clear and concise responses.
- Responses were easy to understand.
- Suitable for simple question-answering tasks.

qwen2.5:3b:
- Produced detailed and informative responses.
- Provided more explanation in some questions.
- Suitable when slightly more detailed answers are required.

Overall:
Both models successfully performed local inference through Ollama.
The response quality differed mainly in detail, explanation style and conciseness.
"""

print(comparison)
```
