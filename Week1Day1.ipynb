```python
import numpy as np
import pandas as pd
```

```python
# 1D Array
array1 = np.array([10, 20, 30, 40, 50])

# 2D Array
array2 = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

# 3D Array
array3 = np.array([
    [[1,2],[3,4]],
    [[5,6],[7,8]]
])

print("1D Array")
print(array1)
print("Shape:", array1.shape)

print("\n2D Array")
print(array2)
print("Shape:", array2.shape)

print("\n3D Array")
print(array3)
print("Shape:", array3.shape)
```

```python
array = np.array([1,2,3,4,5])

result = array + 10

print("Original Array")
print(array)

print("\nAfter Broadcasting")
print(result)
```

```python
a = np.array([1,2,3,4,5])
b = np.array([10,20,30,40,50])

print("Addition")
print(a + b)

print("\nSubtraction")
print(b - a)

print("\nMultiplication")
print(a * b)

print("\nDivision")
print(b / a)
```
