```python
import numpy as np
import pandas as pd
```

```python
# 1D Array
array1 = np.array([10, 20, 30, 40, 50])

# 2D Array
array2 = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

# 3D Array
array3 = np.array([
    [[1,2],[3,4]],
    [[5,6],[7,8]]
])

print("1D Array")
print(array1)
print("Shape:", array1.shape)

print("\n2D Array")
print(array2)
print("Shape:", array2.shape)

print("\n3D Array")
print(array3)
print("Shape:", array3.shape)
```

```python
array = np.array([1,2,3,4,5])

result = array + 10

print("Original Array")
print(array)

print("\nAfter Broadcasting")
print(result)
```

```python
a = np.array([1,2,3,4,5])
b = np.array([10,20,30,40,50])

print("Addition")
print(a + b)

print("\nSubtraction")
print(b - a)

print("\nMultiplication")
print(a * b)

print("\nDivision")
print(b / a)
```

```python
matrix1 = np.array([
    [1,2],
    [3,4]
])

matrix2 = np.array([
    [5,6],
    [7,8]
])

result = np.matmul(matrix1, matrix2)

print("Matrix Multiplication")
print(result)
```

```python
data = {
    "Age":[21,25,28,30,35],
    "Salary":[30000,35000,40000,50000,60000],
    "Experience":[1,3,5,7,10]
}

df = pd.DataFrame(data)

df.to_csv("employee.csv", index=False)

print(df)
```

```python
data = pd.read_csv("employee.csv")

print("Dataset")
print(data)

print("\nMean")
print(data.mean())

print("\nStandard Deviation")
print(data.std())

print("\nCorrelation")
print(data.corr())
```
