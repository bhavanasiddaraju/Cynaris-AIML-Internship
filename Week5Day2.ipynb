```python
import requests

def ask_ollama(prompt, system_prompt, model="llama3.2:3b"):
    url = "http://localhost:11434/api/generate"

    data = {
        "model": model,
        "system": system_prompt,
        "prompt": prompt,
        "stream": False
    }

    response = requests.post(url, json=data)
    response.raise_for_status()

    return response.json()["response"]
```

```python
system_prompt = """
You are a helpful AI tutor for engineering students.
Explain technical concepts in simple language.
Use short explanations and examples when useful.
Avoid unnecessary complexity.
"""
```

```python
prompt1 = "What is Machine Learning?"
prompt2 = "What is a neural network?"

print("PROMPT 1:")
print(ask_ollama(prompt1, system_prompt))

print("\n" + "="*60 + "\n")

print("PROMPT 2:")
print(ask_ollama(prompt2, system_prompt))
```

```python
prompts = [
    "What is overfitting in machine learning?",
    "Explain the difference between supervised and unsupervised learning.",
    "What is a Large Language Model?"
]

for i, prompt in enumerate(prompts, start=3):
    print(f"PROMPT {i}: {prompt}")
    print(ask_ollama(prompt, system_prompt))
    print("\n" + "="*60 + "\n")
```

```python
comparison_questions = [
    "Explain Artificial Intelligence in simple terms.",
    "What is overfitting in machine learning?",
    "Explain the difference between supervised and unsupervised learning."
]

for i, question in enumerate(comparison_questions, start=1):

    print("="*70)
    print(f"QUESTION {i}: {question}")
    print("="*70)

    llama_answer = ask_ollama(
        question,
        system_prompt,
        model="llama3.2:3b"
    )

    qwen_answer = ask_ollama(
        question,
        system_prompt,
        model="qwen2.5:3b"
    )

    print("\nLLAMA 3.2:3B")
    print(llama_answer)

    print("\nQWEN 2.5:3B")
    print(qwen_answer)

    print("\n")
```

```python
results = []

for question in comparison_questions:

    llama_answer = ask_ollama(
        question,
        system_prompt,
        model="llama3.2:3b"
    )

    qwen_answer = ask_ollama(
        question,
        system_prompt,
        model="qwen2.5:3b"
    )

    results.append({
        "Question": question,
        "Llama 3.2 Response": llama_answer,
        "Qwen 2.5 Response": qwen_answer
    })

import pandas as pd

df = pd.DataFrame(results)

df.to_csv("ollama_model_comparison.csv", index=False)

df
```
