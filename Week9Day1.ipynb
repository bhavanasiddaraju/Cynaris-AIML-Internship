```python

from crewai.tools import tool
from ddgs import DDGS

@tool("DuckDuckGo Web Search")
def duckduckgo_web_search(query: str) -> str:
    """Search the web and return relevant page titles, URLs, and snippets."""
    try:
        results = list(DDGS().text(query, max_results=5))

        if not results:
            return "No web results were returned for this query."

        return "\n\n".join(
            f"Title: {item.get('title', 'N/A')}\n"
            f"URL: {item.get('href', 'N/A')}\n"
            f"Snippet: {item.get('body', 'N/A')}"
            for item in results
        )
    except Exception as e:
        return f"Web search error: {type(e).__name__}: {e}"

# Test the tool before connecting it to the crew
print(duckduckgo_web_search.run(
    "Generative AI applications and risks in healthcare"
))
```
