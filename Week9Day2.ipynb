```python

from getpass import getpass
import os

GEMINI_API_KEY = getpass("Enter your Gemini API key: ")

os.environ["GEMINI_API_KEY"] = GEMINI_API_KEY
os.environ["GOOGLE_API_KEY"] = GEMINI_API_KEY

print("Gemini API key configured.")
```

```python

from crewai import LLM

llm = LLM(
    model="gemini/gemini-3.8-flash",
    api_key=GEMINI_API_KEY,
    temperature=0.2
)

print(llm.call("Reply with exactly: W9D2 Gemini test successful"))
```

```python

from crewai import Agent, Task, Crew, Process

researcher = Agent(
    role="Research Analyst",
    goal="Research the benefits, risks, and applications of Generative AI in Education",
    backstory="You are a careful researcher who identifies important facts and evidence.",
    llm=llm,
    verbose=False
)

writer = Agent(
    role="Educational Content Writer",
    goal="Create a clear, structured report from the research findings",
    backstory="You turn technical research into readable, informative reports.",
    llm=llm,
    verbose=False
)

reviewer = Agent(
    role="Quality Reviewer",
    goal="Check accuracy, balance, clarity, and completeness of the report",
    backstory="You critically evaluate reports and suggest practical improvements.",
    llm=llm,
    verbose=False
)

print("All three agents created successfully.")
```

```python

research_task = Task(
    description="Research Generative AI in Education. Identify benefits, risks, examples, and challenges. Topic: {topic}",
    expected_output="A structured research summary covering benefits, risks, examples, and challenges.",
    agent=researcher
)

writing_task = Task(
    description="Write a balanced report on {topic} using the research findings. Include an introduction, benefits, risks, and conclusion.",
    expected_output="A clear report with headings and practical examples.",
    agent=writer
)

review_task = Task(
    description="Review the report for clarity, balance, factual support, and completeness. Provide the improved final report.",
    expected_output="A reviewed report with corrections and recommendations.",
    agent=reviewer
)

print("All three tasks created successfully.")
```

```python

from crewai import LLM

llm = LLM(
    model="gemini/gemini-3.5-flash-lite",
    api_key=GEMINI_API_KEY,
    temperature=0.2
)

print(llm.call("Reply with exactly: Gemini test successful"))
```

```python

researcher.llm = llm
writer.llm = llm
reviewer.llm = llm

print("All three agents updated to the new Gemini model.")
```

```python

crew = Crew(
    agents=[researcher, writer, reviewer],
    tasks=[research_task, writing_task, review_task],
    process=Process.sequential,
    verbose=False
)

import asyncio

initial_result = await crew.kickoff_async(
    inputs={"topic": "Generative AI in Education"}
)

print(initial_result)

```

```python

from ddgs import DDGS
from crewai.tools import tool

@tool("Web Search")
def web_search(query: str) -> str:
    """Search the web for relevant information on a research topic."""
    results = DDGS().text(query, max_results=5)
    return "\n".join(
        f"Title: {r.get('title', '')}\n"
        f"URL: {r.get('href', '')}\n"
        f"Summary: {r.get('body', '')}"
        for r in results
    ) or "No search results found."

print(web_search.run("Generative AI in education benefits and risks"))
```

```python

researcher.tools = [web_search]

web_research_task = Task(
    description="Use the Web Search tool to research {topic}. Find relevant evidence, benefits, risks, and real-world examples. Include source URLs.",
    expected_output="Research findings supported by web search results and source URLs.",
    agent=researcher
)

web_writing_task = Task(
    description="Write a balanced report on {topic} using the research findings and cited URLs.",
    expected_output="A structured report with benefits, risks, examples, and source URLs.",
    agent=writer
)

web_review_task = Task(
    description="Review the report for clarity, balance, evidence, and completeness. Preserve source URLs and avoid unsupported claims.",
    expected_output="An improved, reviewed report.",
    agent=reviewer
)

web_crew = Crew(
    agents=[researcher, writer, reviewer],
    tasks=[web_research_task, web_writing_task, web_review_task],
    process=Process.sequential,
    verbose=False
)

web_result = await web_crew.kickoff_async(
    inputs={"topic": "Generative AI in Education"}
)
print(web_result)
```
